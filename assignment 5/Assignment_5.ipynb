```python
#Q2 - b
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, 3])

def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    frequent_values = values[counts == max_count]

    for value in frequent_values:
        print("Array:", arr)
        print("Most frequent value:", value)
        print("Frequency:", max_count)
        print("Indices:", np.where(arr == value)[0])
        

most_frequent(x)
most_frequent(y)
```

```python
#Q4
Gitansh = np.linspace(10, 100, 25)

print("Array:")
print(Gitansh)

print("Dimensions:", Gitansh.ndim)
print("Shape:", Gitansh.shape)
print("Total elements:", Gitansh.size)
print("Data type:", Gitansh.dtype)
print("Total bytes consumed:", Gitansh.nbytes)

reshaped_Gitansh = Gitansh.reshape(5, 5)

print("Reshaped array:")
print(reshaped_Gitansh)


print("Transpose using T :")
print(reshaped_Gitansh.T)

print("Transpose using transpose():")
print(reshaped_Gitansh.transpose())
```

```python
#Q5
ucs420_gitansh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original 3x4 array:")
print(ucs420_gitansh)

print("Mean:", np.mean(ucs420_gitansh))
print("Median:", np.median(ucs420_gitansh))
print("Maximum:", np.max(ucs420_gitansh))
print("Minimum:", np.min(ucs420_gitansh))
print("Unique elements:", np.unique(ucs420_gitansh))

reshaped_ucs420_gitansh = ucs420_gitansh.reshape(4, 3)

print("Reshaped 4x3 array:")
print(reshaped_ucs420_gitansh)

resized_ucs420_gitansh = ucs420_gitansh.reshape(2, 6)

print("Resized 2x6 array:")
print(resized_ucs420_gitansh)
```
