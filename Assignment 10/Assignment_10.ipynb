```python
# Gitansh
# 1024170268
# Q3

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()
    words = nltk.word_tokenize(text)
    words = [word for word in words if word.isalpha() and word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]
    return words

print(preprocess("High blood pressure can increase the risk of heart disease."))
```

```python
# Gitansh
# 1024170268
# Q4

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

x = [query for query, label in queries]
y = [label for query, label in queries]

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=42, stratify=y
)

model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression(max_iter=1000))
])

model.fit(x_train, y_train)
predictions = model.predict(x_test)

print("accuracy:", accuracy_score(y_test, predictions))
print("precision:", precision_score(y_test, predictions, average="weighted", zero_division=0))
print("recall:", recall_score(y_test, predictions, average="weighted", zero_division=0))
print("f1 score:", f1_score(y_test, predictions, average="weighted", zero_division=0))
print(classification_report(y_test, predictions, zero_division=0))

cm = confusion_matrix(y_test, predictions, labels=model.classes_)
sns.heatmap(cm, annot=True, fmt="d", xticklabels=model.classes_, yticklabels=model.classes_)
plt.xlabel("predicted")
plt.ylabel("actual")
plt.show()
```

```python
# Gitansh
# 1024170268
# Q6

import ipywidgets as widgets
from IPython.display import display, clear_output

emergency_words = [
    "heart attack", "cardiac arrest", "severe chest pain",
    "crushing chest", "cannot breathe", "can't breathe",
    "shortness of breath with chest pain"
]

unsupported_words = ["hair loss", "toothache", "skin care", "knee pain", "eye infection"]

def assistant(query):
    q = query.lower().strip()

    if any(word in q for word in emergency_words):
        return "emergency: this may be a medical emergency. call your local emergency number now. in india, call 112. do not drive yourself. this assistant cannot diagnose you."

    if any(word in q for word in unsupported_words):
        return "this topic is outside the heart health information covered by this assistant. please contact a suitable healthcare professional."

    if len(q.split()) < 2:
        return "low confidence: please enter a more detailed question about heart health."

    probabilities = model.predict_proba([q])[0]
    best_index = probabilities.argmax()
    confidence = probabilities[best_index]
    intent = model.classes_[best_index]

    if confidence < 0.30:
        return "low confidence: i could not identify the intent clearly. please rephrase your heart health question."

    if intent not in knowledge_base:
        return "unsupported query: this assistant does not have information for that topic."

    return "intent: " + intent + "\\n" + knowledge_base[intent] + "\\n\\nplease use this as general information and consult a healthcare professional for personal advice."

test_queries = [
    "what are signs of a heart attack",
    "my chest feels tight",
    "how do i measure bp",
    "how can i lower ldl",
    "how can i prevent heart disease",
    "is walking good for my heart",
    "how can i keep my heart in good shape",
    "tell me about hair loss",
    "maybe",
    "i have severe chest pain and cannot breathe"
]

for q in test_queries:
    print("query:", q)
    print(assistant(q))
    print()

query_box = widgets.Textarea(
    placeholder="enter a heart health question",
    description="query:",
    layout=widgets.Layout(width="90%", height="80px")
)
ask_button = widgets.Button(description="ask")
output = widgets.Output()

def ask_question(button):
    with output:
        clear_output()
        for q in query_box.value.splitlines():
            if q.strip():
                print("query:", q)
                print(assistant(q))
                print()

ask_button.on_click(ask_question)
display(query_box, ask_button, output)
```
