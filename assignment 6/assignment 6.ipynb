```python
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter M: "))

x = np.linspace(-10, 10, 100)

y1 = M*x**2
y2 = M*np.sin(x)

plt.plot(x, y1, label="M*x^2")
plt.plot(x, y2, label="M*sin(x)", linestyle="--")

plt.title("Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

subjects = ["Maths", "Python", "DBMS", "OS", "SE"]
scores = [85, 78, 92, 74, 88]

df = pd.DataFrame({"Subject": subjects, "Score": scores})

sns.barplot(x="Subject", y="Score", data=df)

for i in range(len(scores)):
    plt.text(i, scores[i] + 1, scores[i], ha="center")

plt.title("Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1024170268)

a = np.random.rand(50)
x = np.arange(1, 51)

plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.plot(x, np.cumsum(a))
plt.title("Cumulative Sum")
plt.xlabel("x")
plt.ylabel("sum")
plt.grid()

plt.subplot(2, 2, 2)
plt.scatter(x, a)
plt.title("Random Noise")
plt.xlabel("x")
plt.ylabel("noise")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
df = pd.read_csv(url)

print(df.head())
```

```python
plt.plot(df["month_number"], df["total_profit"], marker="o")
plt.title("Total Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.grid()
plt.show()
```

```python
plt.plot(df["month_number"], df["facecream"], label="Face Cream")
plt.plot(df["month_number"], df["facewash"], label="Face Wash")
plt.plot(df["month_number"], df["toothpaste"], label="Toothpaste")
plt.plot(df["month_number"], df["bathingsoap"], label="Bathing Soap")
plt.plot(df["month_number"], df["shampoo"], label="Shampoo")
plt.plot(df["month_number"], df["moisturizer"], label="Moisturizer")

plt.title("Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
```

```python
df.drop("month_number", axis=1).plot(kind="bar", figsize=(12, 6))

plt.title("All Features")
plt.xlabel("Month")
plt.ylabel("Values")
plt.grid(axis="y")
plt.show()
```
