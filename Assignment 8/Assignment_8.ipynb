```python
# Gitansh
# 1024170268
# Q1

import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from collections import Counter
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = "I like artificial intelligence because it makes computers useful in everyday life. Machine learning helps systems learn from data and improve their results. I enjoy learning how models recognize patterns and make predictions. Python makes it easier to build and test these ideas. I think artificial intelligence will become even more important in the future."

clean_text = re.sub(r'[^\w\s]', '', text.lower())

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print(clean_text)
print(words)
print(sentences)

print("split:", clean_text.split())
print("word tokenize:", words)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print(filtered_words)

freq = Counter(filtered_words)
print(freq)

plt.bar(freq.keys(), freq.values())
plt.xticks(rotation=45)
plt.show()
```

```python
# Gitansh
# 1024170268
# Q2

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = "I like artificial intelligence because it makes computers useful in everyday life. Machine learning helps systems learn from data and improve their results. I enjoy learning how models recognize patterns and make predictions. Python makes it easier to build and test these ideas. I think artificial intelligence will become even more important in the future."

words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
stop_words = set(stopwords.words('english'))
words = [word for word in words if word not in stop_words]

ps = PorterStemmer()
lem = WordNetLemmatizer()

stemmed = [ps.stem(word) for word in words]
lemmatized = [lem.lemmatize(word) for word in words]

print("words:", words)
print("stemmed:", stemmed)
print("lemmatized:", lemmatized)
print("stemming is shorter and faster but can make words less readable")
print("lemmatization gives proper words and is useful when meaning matters")
```

```python
# Gitansh
# 1024170268
# Q3

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

texts = [
    "artificial intelligence helps doctors find diseases early",
    "online shopping makes buying products easy and fast",
    "electric vehicles reduce pollution and use clean energy"
]

cv = CountVectorizer()
bow = cv.fit_transform(texts)

print(pd.DataFrame(bow.toarray(), columns=cv.get_feature_names_out()))

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(texts)
terms = tfidf.get_feature_names_out()

for i in range(len(texts)):
    scores = matrix[i].toarray()[0]
    top = scores.argsort()[-3:][::-1]
    print("text", i + 1)
    print([(terms[j], round(scores[j], 3)) for j in top])
```

```python
# Gitansh
# 1024170268
# Q4

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

ai = "Artificial intelligence allows computers to learn from data. It can recognize patterns and make predictions. AI is used in healthcare, education and transport. It can automate many tasks. AI systems improve with useful training data."

blockchain = "Blockchain stores information in connected digital blocks. It helps people share records securely. Each block is linked with the previous block. Blockchain is used in digital payments and supply chains. It can make records difficult to change."

def clean(text):
    return set(re.findall(r'\b[a-z]+\b', text.lower()))

a = clean(ai)
b = clean(blockchain)

jaccard = len(a.intersection(b)) / len(a.union(b))
print("jaccard similarity:", jaccard)

vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform([ai, blockchain])
cosine = cosine_similarity(matrix[0:1], matrix[1:2])[0][0]
print("cosine similarity:", cosine)

print("cosine similarity gives better insight here because it also considers word importance")
```

```python
# Gitansh
# 1024170268
# Q5

from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "This product is amazing and works really well",
    "I am very happy with the quality and service",
    "The product is useful and easy to use",
    "Very bad experience, the product stopped working",
    "Poor quality and disappointing service",
    "The delivery was late and I did not like it",
    "The product is okay and works as expected"
]

positive = []

for review in reviews:
    blob = TextBlob(review)
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0:
        result = "Positive"
        positive.append(review)
    elif polarity < 0:
        result = "Negative"
    else:
        result = "Neutral"

    print(review)
    print("polarity:", polarity)
    print("subjectivity:", subjectivity)
    print(result)
    print()

cloud = WordCloud(width=800, height=400, background_color="white").generate(" ".join(positive))
plt.imshow(cloud)
plt.axis("off")
plt.show()
```

```python
# Gitansh
# 1024170268
# Q6

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np

text = """Technology is changing the way people live and work. Computers help us solve problems and find information. Learning Python is useful because it can be used for many tasks. Artificial intelligence helps machines learn from examples. Data is important for building good models. Students can use technology to learn new skills and create useful projects. Practice helps people improve their knowledge. New ideas can make the future better."""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1

sequences = []
for line in text.split("."):
    seq = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(seq)):
        sequences.append(seq[:i + 1])

max_len = max(len(x) for x in sequences)
sequences = pad_sequences(sequences, maxlen=max_len, padding="pre")

x = sequences[:, :-1]
y = sequences[:, -1]

model = Sequential()
model.add(Embedding(total_words, 32, input_length=max_len - 1))
model.add(LSTM(50))
model.add(Dense(total_words, activation="softmax"))
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
model.fit(x, y, epochs=100, verbose=0)

def generate(seed, count):
    result = seed
    for i in range(count):
        seq = tokenizer.texts_to_sequences([result])[0]
        seq = pad_sequences([seq], maxlen=max_len - 1, padding="pre")
        prediction = np.argmax(model.predict(seq, verbose=0), axis=-1)[0]
        word = tokenizer.index_word.get(prediction, "")
        result += " " + word
    return result

print(generate("technology", 8))
print(generate("learning", 8))
print(generate("artificial", 8))
```
